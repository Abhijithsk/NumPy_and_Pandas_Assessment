```python
#A
print(df.isnull().sum())

df['Department'] = df['Department'].str.strip().str.title()
num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
print(df.isnull().sum().sum())
```

```python
#B
df['Average_Score'] = df[['Python_Score', 'SQL_Score', 'Pandas_Score', 'NumPy_Score']].mean(axis=1).round(2)
df['Pass_Status'] = df['Average_Score'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')
```

```python
#C
#1
print(df.nlargest(5, 'Average_Score')[['Student_ID', 'Name', 'Department', 'Average_Score']])

#2
print('\n')
dept_avg = df.groupby('Department')['Average_Score'].mean()
print(dept_avg.idxmax(), round(dept_avg.max(), 2))

#3
print('\n')
print(df.groupby('Department')['Attendance'].mean().round(2))


#4
print('\n')
print(df['Pass_Status'].value_counts())


```

```python
#D
summary = df.groupby('Department').agg(
    Total_Students=('Student_ID', 'count'),
    Average_Score=('Average_Score', 'mean'),
    Average_Attendance=('Attendance', 'mean')).round(2)

counts = pd.crosstab(df['Department'], df['Pass_Status'])
summary['Pass_Count'] = counts['Pass']
summary['Fail_Count'] = counts['Fail']

print(summary.reset_index())
```

```python
#A
df = pd.read_csv('/content/drive/MyDrive/DSAB13/Class_Activity/assesment2/sales.csv')
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', dayfirst=True)
df['Sales'] = pd.to_numeric(df['Sales'].str.replace(',', ''))
df['Discount'] = df['Discount'].str.replace('%', '').astype(float)
df.loc[df['Discount'] > 1, 'Discount'] = df['Discount'] / 100

for col in ['Product', 'Category', 'Region']:
    df[col] = df[col].str.strip().str.title()

print(df.isnull().sum())

df['Region'] = df['Region'].fillna('Unknown')
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
df['Profit'] = df['Profit'].fillna(df['Profit'].median())
```

```python
#B
print(df['Sales'].sum())
print(df['Profit'].sum())
print(df['Sales'].mean())
print(df.groupby('Product')['Sales'].sum().sort_values(ascending=False).head(5))
print(df.groupby('Region')['Sales'].sum().idxmax())
```

```python
#C
summary = df.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum'),
    Average_Discount=('Discount', 'mean')).round(2)

summary = summary.sort_values('Total_Sales', ascending=False)
print(summary)
```
